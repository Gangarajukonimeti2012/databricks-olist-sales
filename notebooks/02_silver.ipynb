# 02 Silver: clean the data
One query per table: keep the useful columns, remove duplicates, drop rows with no ID.

```python
%sql
CREATE OR REPLACE TABLE olist.silver.orders AS
SELECT DISTINCT
       order_id,
       customer_id,
       order_status,
       CAST(order_purchase_timestamp AS DATE)      AS order_date,
       CAST(order_delivered_customer_date AS DATE) AS delivered_date
FROM olist.bronze.orders
WHERE order_id IS NOT NULL
```

```python
%sql
CREATE OR REPLACE TABLE olist.silver.order_items AS
SELECT DISTINCT order_id, order_item_id, product_id, price, freight_value
FROM olist.bronze.order_items
WHERE order_id IS NOT NULL
```

```python
%sql
CREATE OR REPLACE TABLE olist.silver.customers AS
SELECT DISTINCT customer_id, customer_city, customer_state
FROM olist.bronze.customers
WHERE customer_id IS NOT NULL
```

```python
%sql
CREATE OR REPLACE TABLE olist.silver.products AS
SELECT DISTINCT product_id, product_category_name AS product_category
FROM olist.bronze.products
WHERE product_id IS NOT NULL
```

```python
%sql
CREATE OR REPLACE TABLE olist.silver.payments AS
SELECT order_id, payment_type, payment_value
FROM olist.bronze.payments
WHERE order_id IS NOT NULL
```
