# 01 Bronze: load the CSV files into tables
Read each CSV and save it as a table. No cleaning.

The `*` in a file name also picks up files uploaded in parts (`..._part1.csv`, `..._part2.csv`).

```python
df = spark.read.csv("/Volumes/olist/bronze/raw/olist_orders_dataset*.csv", header=True, inferSchema=True)
df.write.mode("overwrite").saveAsTable("olist.bronze.orders")
```

```python
df = spark.read.csv("/Volumes/olist/bronze/raw/olist_order_items_dataset*.csv", header=True, inferSchema=True)
df.write.mode("overwrite").saveAsTable("olist.bronze.order_items")
```

```python
df = spark.read.csv("/Volumes/olist/bronze/raw/olist_customers_dataset*.csv", header=True, inferSchema=True)
df.write.mode("overwrite").saveAsTable("olist.bronze.customers")
```

```python
df = spark.read.csv("/Volumes/olist/bronze/raw/olist_products_dataset*.csv", header=True, inferSchema=True)
df.write.mode("overwrite").saveAsTable("olist.bronze.products")
```

```python
df = spark.read.csv("/Volumes/olist/bronze/raw/olist_order_payments_dataset*.csv", header=True, inferSchema=True)
df.write.mode("overwrite").saveAsTable("olist.bronze.payments")
```
