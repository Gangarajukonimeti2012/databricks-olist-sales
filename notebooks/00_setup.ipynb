# 00 Setup (run once)
Creates the catalog, one schema per layer, and a volume for the CSV files.

```python
%sql
CREATE CATALOG IF NOT EXISTS olist
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS olist.bronze
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS olist.silver
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS olist.gold
```

```python
%sql
CREATE VOLUME IF NOT EXISTS olist.bronze.raw
```

Now upload the CSV files: Catalog → `olist` → `bronze` → `raw` → **Upload to this volume**.
