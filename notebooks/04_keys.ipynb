# 04 Keys: star schema relationships in Unity Catalog
Run after `03_gold`. A primary key column must be NOT NULL, so that is set first.

## Primary keys

```python
%sql
ALTER TABLE olist.gold.dim_customer ALTER COLUMN customer_id SET NOT NULL
```

```python
%sql
ALTER TABLE olist.gold.dim_customer ADD CONSTRAINT pk_customer PRIMARY KEY (customer_id)
```

```python
%sql
ALTER TABLE olist.gold.dim_product ALTER COLUMN product_id SET NOT NULL
```

```python
%sql
ALTER TABLE olist.gold.dim_product ADD CONSTRAINT pk_product PRIMARY KEY (product_id)
```

```python
%sql
ALTER TABLE olist.gold.dim_date ALTER COLUMN order_date SET NOT NULL
```

```python
%sql
ALTER TABLE olist.gold.dim_date ADD CONSTRAINT pk_date PRIMARY KEY (order_date)
```

## Foreign keys

```python
%sql
ALTER TABLE olist.gold.fact_orders ADD CONSTRAINT fk_customer
FOREIGN KEY (customer_id) REFERENCES olist.gold.dim_customer (customer_id)
```

```python
%sql
ALTER TABLE olist.gold.fact_orders ADD CONSTRAINT fk_product
FOREIGN KEY (product_id) REFERENCES olist.gold.dim_product (product_id)
```

```python
%sql
ALTER TABLE olist.gold.fact_orders ADD CONSTRAINT fk_date
FOREIGN KEY (order_date) REFERENCES olist.gold.dim_date (order_date)
```
